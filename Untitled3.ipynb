```python
%%writefile main.py
import streamlit as st
import numpy as np
import tempfile
import os
import torch
import torch.nn.functional as F
from PIL import Image
import matplotlib.pyplot as plt

st.set_page_config(page_title="VITAL VISION", layout="wide", page_icon="🏥")
st.title("🏥 VITAL VISION: Explainable Diagnostic Assistant")

st.sidebar.title("Clinical Modules")
module_selection = st.sidebar.radio(
    "Select Imaging Modality",
    ["Chest X-Ray Pathology Screening", "CT/MRI 3D Anatomical Segmentation"]
)

# ---------------------------------------------------------
# PyTorch Grad-CAM Implementation
# ---------------------------------------------------------
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None

        # Register PyTorch hooks to capture activations and gradients
        self.forward_hook = target_layer.register_forward_hook(self.save_activation)
        self.backward_hook = target_layer.register_backward_hook(self.save_gradient)

    def save_activation(self, module, input, output):
        self.activations = output

    def save_gradient(self, module, grad_input, grad_output):
        self.gradients = grad_output[0]

    def generate_heatmap(self, x, class_idx):
        self.model.zero_grad()
        output = self.model(x)

        # Get gradient of the target pathology's confidence score
        score = output[0, class_idx]
        score.backward()

        # Global Average Pooling of gradients
        gradients = self.gradients.cpu().data.numpy()[0]
        activations = self.activations.cpu().data.numpy()[0]

        weights = np.mean(gradients, axis=(1, 2))
        cam = np.zeros(activations.shape[1:], dtype=np.float32)

        # Weighted sum of feature maps
        for i, w in enumerate(weights):
            cam += w * activations[i]

        # Apply ReLU to focus only on positive feature influence
        cam = np.maximum(cam, 0)

        # Resize to input tensor size (224x224) using PyTorch bilinear interpolation
        cam_tensor = torch.from_numpy(cam).unsqueeze(0).unsqueeze(0)
        cam_resized = F.interpolate(cam_tensor, size=(224, 224), mode='bilinear', align_corners=False)
        cam_resized = cam_resized.squeeze().numpy()

        # Normalize heatmap to [0, 1] range
        if np.max(cam_resized) != np.min(cam_resized):
            cam_resized = (cam_resized - np.min(cam_resized)) / (np.max(cam_resized) - np.min(cam_resized))
        else:
            cam_resized = np.zeros_like(cam_resized)

        return cam_resized, output.detach().cpu().numpy()[0]

    def release_hooks(self):
        self.forward_hook.remove()
        self.backward_hook.remove()

# ---------------------------------------------------------
# 2D Image Preprocessing
# ---------------------------------------------------------
def preprocess_xray(file_obj, filename):
    import torchxrayvision as xrv

    if filename.lower().endswith('.dcm'):
        import pydicom
        dicom = pydicom.dcmread(file_obj)
        img_array = dicom.pixel_array.astype(float)
        if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
            img_array = np.max(img_array) - img_array
    else:
        image = Image.open(file_obj)
        img_array = np.array(image).astype(float)

    if len(img_array.shape) > 2:
        img_array = img_array.mean(axis=2)

    img_norm = xrv.datasets.normalize(img_array, np.max(img_array))

    h, w = img_norm.shape
    max_dim = max(h, w)
    pad_h = (max_dim - h) // 2
    pad_w = (max_dim - w) // 2

    padded_img = np.pad(
        img_norm,
        ((pad_h, max_dim - h - pad_h), (pad_w, max_dim - w - pad_w)),
        mode='constant',
        constant_values=img_norm.min()
    )

    padded_tensor = torch.from_numpy(padded_img).unsqueeze(0).unsqueeze(0).float()
    resized_tensor = F.interpolate(
        padded_tensor, size=(224, 224), mode='bilinear', align_corners=False
    )

    return resized_tensor, img_norm

# ---------------------------------------------------------
# Interface Modality Mapping
# ---------------------------------------------------------
if module_selection == "Chest X-Ray Pathology Screening":
    st.header("Chest X-Ray Pathology Screening")
    st.write("Upload a 16-bit hospital `.dcm` file or `.png`/`.jpg` chest scan.")

    uploaded_file = st.file_uploader("Upload Medical Scan", type=['dcm', 'png', 'jpg', 'jpeg'])

    if uploaded_file is not None:
        col1, col2 = st.columns([1, 1])

        with col1:
            st.subheader("Input Scan")
            try:
                tensor_input, raw_norm = preprocess_xray(uploaded_file, uploaded_file.name)

                # Convert back to uint8 3-channel grayscale for visualization
                norm_img = (raw_norm - raw_norm.min()) / (raw_norm.max() - raw_norm.min())
                display_img = Image.fromarray((norm_img * 255).astype(np.uint8)).convert("RGB")
                display_img_resized = display_img.resize((224, 224))

                st.image(display_img_resized, caption="Patient Scan (Resized)", use_column_width=True)
            except Exception as e:
                st.error(f"Preprocessing error: {str(e)}")
                st.stop()

        with col2:
            st.subheader("Visual Explanations (Grad-CAM)")
            with st.spinner("Analyzing scans & activating saliency maps..."):
                try:
                    import torchxrayvision as xrv
                    model = xrv.models.DenseNet(weights="densenet121-res224-all")
                    model.eval()

                    # Target the final Convolutional Block of DenseNet
                    target_layer = model.features[-1]
                    grad_cam = GradCAM(model, target_layer)

                    # Compute pathology metrics (Default target: top prediction)
                    dummy_idx = 0
                    heatmap, outputs = grad_cam.generate_heatmap(tensor_input, dummy_idx)

                    # Map scores back to clinical tags
                    results = dict(zip(model.pathologies, outputs))
                    sorted_results = dict(sorted(results.items(), key=lambda item: item[1], reverse=True))

                    # Let user choose which pathology's map they want to inspect!
                    selected_pathology = st.selectbox(
                        "Inspect Saliency Map For:",
                        list(sorted_results.keys())
                    )

                    pathology_idx = model.pathologies.index(selected_pathology)

                    # Regenerate heatmap for user-selected pathology
                    heatmap, _ = grad_cam.generate_heatmap(tensor_input, pathology_idx)
                    grad_cam.release_hooks()

                    # Overlay using matplotlib jet colormap
                    img_np = np.array(display_img_resized) / 255.0
                    jet_heatmap = plt.cm.jet(heatmap)[:, :, :3]
                    overlay = 0.6 * img_np + 0.4 * jet_heatmap

                    st.image(overlay, caption=f"Attribution Map for: {selected_pathology}", use_column_width=True)

                    # Display metrics with confidence feedback
                    score = sorted_results[selected_pathology] * 100
                    if score > 50:
                        st.error(f"**{selected_pathology}**: {score:.1f}% Match (High Severity Risk)")
                    elif score > 25:
                        st.warning(f"**{selected_pathology}**: {score:.1f}% Match (Borderline Finding)")
                    else:
                        st.success(f"**{selected_pathology}**: {score:.1f}% Match (WNL - Within Normal Limits)")

                except Exception as e:
                    st.error
```
